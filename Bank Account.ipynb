```python
from datetime import datetime

class BankAccount:
    def __init__(self, holder_name: str, account_number: str, account_type: str, initial_balance: float = 0.0):
        """
        Creates an Account.
        Initializes the holder name, account number, type, balance, and transaction history.
        """
        self._holder_name = holder_name
        self._account_number = account_number
        self._account_type = account_type
        self._balance = max(0.0, initial_balance)
        self._transaction_history = []
        
        # Record the account transaction
        self._log_transaction("Account Created", self._balance)


    def deposit(self, amount: float) -> bool:
        """Deposits Money into the account."""
        if amount <= 0:
            print("Deposit amount must be greater than zero.")
            return False
        
        self._balance += amount
        self._log_transaction("Deposit", amount)
        print(f"Successfully deposited ${amount:,.2f}. New Balance: ${self._balance:,.2f}")
        return True

    def withdraw(self, amount: float) -> bool:
        """Withdrawing Money from the account."""
        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")
            return False
        
        if amount > self._balance:
            print(f"Insufficient funds. Available balance: ${self._balance:,.2f}")
            return False
        
        self._balance -= amount
        self._log_transaction("Withdrawal", amount)
        print(f"Successfully withdrew ${amount:,.2f}. New Balance: ${self._balance:,.2f}")
        return True

    def check_balance(self) -> float:
        """Checking the balance."""
        print(f"Current Balance: ${self._balance:,.2f}")
        return self._balance

    # --- Getter Methods ---

    def get_holder_name(self) -> str:
        """Getting the Holder name."""
        return self._holder_name

    def get_account_number(self) -> str:
        """Getting the Account number."""
        return self._account_number

    def get_account_type(self) -> str:
        """Getting the Account type."""
        return self._account_type

    def get_transaction_history(self) -> list:
        """Keeping and returning a transaction history."""
        print(f"\n--- Transaction History for Account {self._account_number} ---")
        if not self._transaction_history:
            print("No transactions recorded.")
        for tx in self._transaction_history:
            print(f"[{tx['timestamp']}] {tx['type']}: {tx['amount_str']} | Balance: {tx['balance_str']}")
        return self._transaction_history

    # --- Private Helper Methods ---

    def _log_transaction(self, action_type: str, amount: float):
        """Internal helper to timestamp and append transactions to history."""
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        self._transaction_history.append({
            "timestamp": timestamp,
            "type": action_type,
            "amount": amount,
            "amount_str": f"${amount:,.2f}",
            "balance_str": f"${self._balance:,.2f}"
        })
if __name__ == "__main__":


    #
    # 1. Create Account
    print("=== Creating Account ===")
    account = BankAccount(
        holder_name="Bubba Bo Bob", 
        account_number="31415926", 
        account_type="Savings", 
        initial_balance=500.0
    )

    # 2. Getting Information
    print("\n=== Account Meta Info ===")
    print(f"Holder Name: {account.get_holder_name()}")
    print(f"Account Number: {account.get_account_number()}")
    print(f"Account Type: {account.get_account_type()}")

    # 3. Banking Operations
    print("\n=== Transactions ===")
    account.check_balance()
    account.deposit(250.50)
    account.withdraw(100.00)
    account.withdraw(800.00)  # Should fail due to lack of funds
    
    # 4. Extracting History
    account.get_transaction_history()
```
